```python
from google.colab import files
import os, re, zipfile

print('WARNING: All existing results files will be deleted from the file tab upon uploading a new zip file.')
zip = files.upload()
zip_path = os.path.abspath(next(iter(zip)))

pat = re.compile(r"(?:^|/)OpenSimData/Kinematics/([^/]+)\.mot$")
with zipfile.ZipFile(zip_path) as z:
    trials = sorted({m.group(1) for n in z.namelist() if (m := pat.search(n))})
print("Trials found:", trials, f"/n{trials[0]} will be used for analysis.")
trial = trials[0]   # to use a different trial, change this line

!cd repo/AdditionalProcessing && rm -rf results && ../bin/micromamba run -n opencap python -u colab_run.py "{zip_path}" "{trial}"
```
